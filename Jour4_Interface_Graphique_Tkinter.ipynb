## 1. Les bases de Tkinter

Tkinter est la bibliothèque graphique **incluse par défaut** avec Python, aucune installation nécessaire. C'est pour ça qu'elle est très utilisée pour des outils internes en entreprise (petits logiciels métier, sans dépendances lourdes).

### Concepts clés

| Concept | Rôle |
|---|---|
| `Tk()` | La fenêtre principale de l'application |
| `Widget` | Un élément graphique : bouton, champ de texte, étiquette... |
| `pack()` / `grid()` | Les gestionnaires de disposition (placement des widgets) |
| `mainloop()` | La boucle qui garde la fenêtre ouverte et réactive |

**Points clés à retenir :**

## 3. Afficher une liste de données avec `Treeview`

Pour afficher une liste de clients façon tableau (comme dans un vrai logiciel de gestion), on utilise le widget `Treeview` du sous-module `ttk`.

````python
%%writefile liste_treeview.py
import tkinter as tk
from tkinter import ttk

# Données de démonstration (viendront de GestionCommerciale demain)
clients_demo = [
    (1, "Amina Traoré", "amina@example.com", "0612345678"),
    (2, "Jean Kouassi", "jean.k@example.com", "0623456789"),
]

fenetre = tk.Tk()
fenetre.title("Liste des clients")
fenetre.geometry("500x200")

colonnes = ("id", "nom", "email", "telephone")
tableau = ttk.Treeview(fenetre, columns=colonnes, show="headings")

# En-têtes de colonnes
tableau.heading("id", text="ID")
tableau.heading("nom", text="Nom")
tableau.heading("email", text="Email")
tableau.heading("telephone", text="Téléphone")

tableau.column("id", width=40)
tableau.column("nom", width=150)
tableau.column("email", width=180)
tableau.column("telephone", width=100)

# Remplissage avec les données
for client in clients_demo:
    tableau.insert("", tk.END, values=client)

tableau.pack(padx=10, pady=10, fill="both", expand=True)

fenetre.mainloop()
````

## 4. Interface complète connectée à `GestionCommerciale`

On assemble maintenant tout : le formulaire, la liste, et **la vraie logique métier du Jour 3**. C'est ce fichier qui deviendra la base du mini-projet de fin de semaine.

**Principe d'architecture important :** notez que `GestionCommerciale` ne contient **aucune ligne de code Tkinter**. L'interface *appelle* la logique métier, elle ne s'y mélange pas. C'est ce qui permettrait, plus tard, de faire une version web sans réécrire tout le moteur métier.

````python
%%writefile gestion_commerciale.py
"""
Module métier : logique de gestion des clients et commandes.
Ce module ne dépend d'AUCUNE bibliothèque graphique : il peut être
réutilisé par une interface Tkinter, une interface web, ou des tests automatisés.
"""
import re


def valider_email(email):
    motif = r"^[\w.\-]+@[\w\-]+\.[a-zA-Z]{2,}$"
    return re.match(motif, email) is not None


def valider_telephone(telephone):
    return telephone.isdigit() and len(telephone) == 10


class Client:
    def __init__(self, id_client, nom, email, telephone):
        if not nom or not nom.strip():
            raise ValueError("Le nom du client est obligatoire.")
        if not valider_email(email):
            raise ValueError(f"Email invalide : {email}")
        if not valider_telephone(telephone):
            raise ValueError(f"Téléphone invalide : {telephone}")

        self.id_client = id_client
        self.nom = nom.strip()
        self.email = email
        self.telephone = telephone
        self.commandes = []

    def ajouter_commande(self, commande):
        self.commandes.append(commande)

    def total_depense(self):
        return round(sum(c.total() for c in self.commandes), 2)

    def __repr__(self):
        return f"Client(id={self.id_client}, nom='{self.nom}')"


class Commande:
    def __init__(self, id_commande, client, date, produit, quantite, prix_unitaire):
        if quantite <= 0:
            raise ValueError("La quantité doit être strictement positive.")
        if prix_unitaire < 0:
            raise ValueError("Le prix ne peut pas être négatif.")

        self.id_commande = id_commande
        self.client = client
        self.date = date
        self.produit = produit
        self.quantite = quantite
        self.prix_unitaire = prix_unitaire

    def total(self):
        return round(self.quantite * self.prix_unitaire, 2)

    def __repr__(self):
        return f"Commande(id={self.id_commande}, produit='{self.produit}')"


class GestionCommerciale:
    def __init__(self):
        self._clients = {}
        self._prochain_id_client = 1
        self._prochain_id_commande = 1

    def ajouter_client(self, nom, email, telephone):
        for client in self._clients.values():
            if client.email == email:
                raise ValueError(f"Un client avec l'email {email} existe déjà.")

        client = Client(self._prochain_id_client, nom, email, telephone)
        self._clients[client.id_client] = client
        self._prochain_id_client += 1
        return client

    def ajouter_commande(self, id_client, date, produit, quantite, prix_unitaire):
        if id_client not in self._clients:
            raise ValueError(f"Client inconnu (id={id_client}).")

        client = self._clients[id_client]
        commande = Commande(self._prochain_id_commande, client, date, produit, quantite, prix_unitaire)
        client.ajouter_commande(commande)
        self._prochain_id_commande += 1
        return commande

    def lister_clients(self):
        return list(self._clients.values())

    def obtenir_client(self, id_client):
        return self._clients.get(id_client)

    def meilleur_client(self):
        if not self._clients:
            return None
        return max(self._clients.values(), key=lambda c: c.total_depense())
````

Maintenant, l'interface graphique qui **importe** ce module :

````python
%%writefile interface_gestion.py
"""
Interface graphique Tkinter pour l'application de gestion commerciale.
Lancer avec : python interface_gestion.py
"""
import tkinter as tk
from tkinter import ttk, messagebox
from gestion_commerciale import GestionCommerciale

class ApplicationGestion:
    """Fenêtre principale de l'application."""

    def __init__(self, fenetre):
        self.gestion = GestionCommerciale()  # le moteur métier du Jour 3
        self.fenetre = fenetre
        self.fenetre.title("Gestion Clients / Commandes")
        self.fenetre.geometry("600x450")

        self._construire_formulaire()
        self._construire_tableau()

    def _construire_formulaire(self):
        cadre_formulaire = tk.LabelFrame(self.fenetre, text="Ajouter un client", padx=10, pady=10)
        cadre_formulaire.pack(padx=10, pady=10, fill="x")

        tk.Label(cadre_formulaire, text="Nom :").grid(row=0, column=0, sticky="e", padx=5, pady=5)
        self.champ_nom = tk.Entry(cadre_formulaire, width=25)
        self.champ_nom.grid(row=0, column=1, padx=5, pady=5)

        tk.Label(cadre_formulaire, text="Email :").grid(row=1, column=0, sticky="e", padx=5, pady=5)
        self.champ_email = tk.Entry(cadre_formulaire, width=25)
        self.champ_email.grid(row=1, column=1, padx=5, pady=5)

        tk.Label(cadre_formulaire, text="Téléphone :").grid(row=2, column=0, sticky="e", padx=5, pady=5)
        self.champ_telephone = tk.Entry(cadre_formulaire, width=25)
        self.champ_telephone.grid(row=2, column=1, padx=5, pady=5)

        bouton = tk.Button(cadre_formulaire, text="Ajouter", command=self.ajouter_client)
        bouton.grid(row=3, column=0, columnspan=2, pady=10)

    def _construire_tableau(self):
        cadre_tableau = tk.LabelFrame(self.fenetre, text="Liste des clients", padx=10, pady=10)
        cadre_tableau.pack(padx=10, pady=10, fill="both", expand=True)

        colonnes = ("id", "nom", "email", "telephone", "total_depense")
        self.tableau = ttk.Treeview(cadre_tableau, columns=colonnes, show="headings")

        libelles = {"id": "ID", "nom": "Nom", "email": "Email",
                    "telephone": "Téléphone", "total_depense": "Total dépensé (€)"}
        for col in colonnes:
            self.tableau.heading(col, text=libelles[col])
            self.tableau.column(col, width=110)

        self.tableau.pack(fill="both", expand=True)

    def ajouter_client(self):
        """Callback du bouton 'Ajouter' : lit le formulaire, appelle la logique métier, met à jour l'affichage."""
        nom = self.champ_nom.get()
        email = self.champ_email.get()
        telephone = self.champ_telephone.get()

        try:
            # Toute la validation est déléguée à la classe métier : l'interface ne fait QUE l'affichage
            self.gestion.ajouter_client(nom, email, telephone)
        except ValueError as erreur:
            messagebox.showerror("Erreur de saisie", str(erreur))
            return

        self._rafraichir_tableau()
        self.champ_nom.delete(0, tk.END)
        self.champ_email.delete(0, tk.END)
        self.champ_telephone.delete(0, tk.END)
        messagebox.showinfo("Succès", "Client ajouté avec succès.")

    def _rafraichir_tableau(self):
        """Vide et reconstruit le tableau à partir de l'état actuel de la logique métier."""
        for ligne in self.tableau.get_children():
            self.tableau.delete(ligne)

        for client in self.gestion.lister_clients():
            self.tableau.insert("", tk.END, values=(
                client.id_client, client.nom, client.email,
                client.telephone, client.total_depense()
            ))


if __name__ == "__main__":
    fenetre = tk.Tk()
    app = ApplicationGestion(fenetre)
    fenetre.mainloop()
````
