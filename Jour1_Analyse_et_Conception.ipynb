## 1. Le cycle de vie d'un projet informatique

En entreprise, un projet applicatif suit ces étapes :

| Étape | Ce qu'on fait | Ce qu'on produit |
|---|---|---|
| 1. Analyse du besoin | Comprendre ce que veut le client/utilisateur | Cahier des charges |
| 2. Conception | Modéliser la solution | Diagrammes (classes, MCD, maquettes) |
| 3. Développement | Écrire le code | Le programme |
| 4. Tests | Vérifier que ça marche | Rapport de tests |
| 5. Déploiement | Mettre en production | Application livrée |
| 6. Maintenance | Corriger, faire évoluer | Mises à jour |


### Pourquoi c'est important de ne pas sauter l'analyse ?

Un client qui dit *"je veux gérer mes clients"* ne vous donne pas un cahier des charges, il vous donne une intention. Votre travail est de la transformer en besoins précis, par des questions :

- Qu'est-ce qu'un "client" pour cette entreprise ? (nom, entreprise, email, téléphone... ?)
- Que doit-on pouvoir faire avec une commande ? (créer, modifier, annuler... ?)
- Qui va utiliser l'application ? (1 personne, plusieurs ?)
- Combien de clients/commandes en volume approximatif ?
- Y a-t-il des règles métier ? (ex : pas de commande sans client existant)

## 4. Conception : Identifier les entités et leurs relations

Avant de coder, on **modélise**. Deux outils simples et très utilisés en entreprise, même pour de petits projets :

### a) Le diagramme de classes (simplifié, façon UML)

```
┌─────────────────────────┐
│         Client          │
├─────────────────────────┤
│ - id_client : int       │
│ - nom : str             │
│ - email : str           │
│ - telephone : str       │
├─────────────────────────┤
│ + ajouter_commande()    │
│ + total_depense()       │
└─────────────────────────┘
            │ 1
            │
            │ 0..*
┌─────────────────────────┐
│        Commande         │
├─────────────────────────┤
│ - id_commande : int     │
│ - client : Client       │
│ - date : str            │
│ - produit : str         │
│ - quantite : int        │
│ - prix_unitaire : float │
├─────────────────────────┤
│ + total() : float       │
└─────────────────────────┘
```

**Lecture du diagramme :**
- Chaque **rectangle** = une classe (un "moule" à objets)
- Les **attributs** (données) sont dans la 2ᵉ section, précédés de `-` (privé)
- Les **méthodes** (actions) sont dans la 3ᵉ section, précédées de `+` (public)
- La ligne entre les deux classes indique la **relation** : *"1 Client peut avoir 0 à plusieurs Commandes"*

### b) Le Modèle Conceptuel de Données (MCD) 

```
CLIENT (id_client, nom, email, telephone)
COMMANDE (id_commande, date, produit, quantite, prix_unitaire, #id_client)
```
Le `#` indique une **clé étrangère** : chaque commande référence le client qui l'a passée.

## 5. Traduire la conception en squelette Python

En entreprise, une fois le diagramme validé (souvent avec le chef de projet ou le client), on écrit le **squelette de code** : les classes avec leurs attributs, mais sans encore développer toute la logique. Cela permet de valider la structure avant d'investir du temps de développement.


````python
# Squelette du projet fil rouge 

class Client:
    """Représente un client de l'entreprise."""
    def __init__(self, id_client, nom, email, telephone):
        self.id_client = id_client
        self.nom = nom
        self.email = email
        self.telephone = telephone
        self.commandes = []  # une liste vide

    def __repr__(self):
        return f"Client({self.id_client}, {self.nom})"


class Commande:
    """Représente une commande passée par un client."""
    def __init__(self, id_commande, client, date, produit, quantite, prix_unitaire):
        self.id_commande = id_commande
        self.client = client
        self.date = date
        self.produit = produit
        self.quantite = quantite
        self.prix_unitaire = prix_unitaire

    def __repr__(self):
        return f"Commande({self.id_commande}, {self.produit})"


# Test rapide du squelette
client_test = Client(1, "Amina Traoré", "amina@example.com", "0612345678")
print(client_test)
print("Commandes de ce client (vide pour l'instant) :", client_test.commandes)
````
